```python
from sklearn.svm import SVC
from model_pipeline import model_pipeline
import pandas as pd

df = pd.read_csv("../feature_engineering/data_with_new_features.csv")

results_svc = model_pipeline(
    df,
    model_factory=lambda **p: SVC(
        random_state=42,
        probability=True,      # so you can get predict_proba if you need it
        **p
    ),
    normalize='global',
    test_val_split=0.1,
    param_distributions={
        'clf__C':           (0.1, 10.0, 0.1),      # inverse regularization strength
        'clf__kernel':      ['linear','rbf'],      # try linear vs. RBF
        'clf__gamma':       ['scale','auto'],      # kernel bandwidth
    },
    n_trials=30,
    search_cv=3,
    test_participant='clara',
    use_rfe=False
)
print(results_svc['reports']['clara']['f1_weighted'])
```
